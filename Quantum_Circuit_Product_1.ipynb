## Objectives and mathematical background
A qubit is $|psi> = alpha|0> + beta|1>$ with $|alpha|^2 + |beta|^2 = 1$. Measurement returns 0 or 1 according to the squared amplitudes. The Hadamard gate creates superposition, controlled gates create correlations, and measurement converts quantum information to classical outcomes.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

PROJECT = Path.cwd()
if (PROJECT / 'circuits.py').exists() is False:
    PROJECT = PROJECT / 'Quantum-Circuit-Simulator'
sys.path.insert(0, str(PROJECT))

from circuits import catalog, bell_state, ghz_state, hadamard, pauli_x, rotation
from simulator import measure, statevector, transpilation_summary
from visualization import bloch, histogram
```

## Complete circuit catalog

```python
for name, builder in catalog().items():
    print(f'\n{name}')
    print(builder().draw(output='text'))
```

## Hadamard state and measurement

```python
h_circuit = hadamard()
print(h_circuit.draw(output='text'))
h_state = statevector(h_circuit)
print('Statevector:', h_state)
h_counts = measure(h_circuit, shots=1024)
print('Counts:', h_counts)
histogram(h_counts)
plt.show()
bloch(h_state)
plt.show()
```

## Bell and GHZ entanglement results

```python
for name, circuit in [('Bell', bell_state()), ('GHZ', ghz_state())]:
    print(name)
    print(circuit.draw(output='text'))
    print('Statevector:', statevector(circuit))
    print('Counts:', measure(circuit, shots=1024))
```

## Theoretical versus simulated comparison

```python
comparison = {
    'Pauli-X': {'theory': {'1': 1.0}, 'simulation': measure(pauli_x())},
    'Bell': {'theory': {'00': 0.5, '11': 0.5}, 'simulation': measure(bell_state())},
    'GHZ': {'theory': {'000': 0.5, '111': 0.5}, 'simulation': measure(ghz_state())},
}
comparison
```

## Rotation and transpilation analysis

```python
rotation_circuit = rotation('y', 3.14159 / 2)
print(rotation_circuit.draw(output='text'))
print(transpilation_summary(rotation_circuit))
```

## Conclusions and limitations
The ideal simulator reproduces the expected basis-state, superposition, and entanglement behavior. Counts vary statistically with finite shots. Real-device noise, connectivity, calibration, and queue time are outside the base submission but are natural future enhancements.
